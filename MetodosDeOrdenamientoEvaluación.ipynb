# Comparación de Algoritmos de Ordenamiento: Burbuja vs Quicksort

## **Objetivo**
Implementar y evaluar los algoritmos de ordenamiento **Burbuja** y **Quicksort** en Python. Se medirá su rendimiento con diferentes tamaños de datos para analizar cuál es más eficiente.

## **Metodología**
1. Implementar los algoritmos en Python.
2. Evaluar su rendimiento usando `time` y `timeit`.
3. Comparar los tiempos y analizar los resultados.

# 1. Implementación del algoritmo de ordenamiento burbuja


```python
import time
import random

def bubble_sort(arr):
    """Implementación del algoritmo de ordenamiento Burbuja."""
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]  # Intercambio de elementos
```

# 2. Implementacion del algoritmo de ordenamiento QuickSort


```python
def quicksort(arr):
    """Implementación del algoritmo de ordenamiento Quicksort."""
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]  # Elegimos el pivote como el elemento central
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)
```

# 3. Pruebas de rendimeinto 

## 3.1 Medición del tiempo de ejecución del método Burbuja

```python
# Generar lista aleatoria de 1000 elementos
arr = [random.randint(0, 1000) for _ in range(1000)]

# Medir tiempo de ejecución del algoritmo Burbuja
start_time = time.time()
bubble_sort(arr.copy())  # Usamos una copia para evitar modificar la original
end_time = time.time()

print(f"Tiempo de ejecución Burbuja: {end_time - start_time:.6f} segundos")
```

# 4. Comparación con diferentes tamaños de datos

```python
import timeit

sizes = [100, 500, 1000, 5000]  # Diferentes tamaños de listas

for size in sizes:
    arr = [random.randint(0, 10000) for _ in range(size)]
    
    # Medir tiempo Burbuja
    bubble_time = timeit.timeit(lambda: bubble_sort(arr.copy()), number=1)
    
    # Medir tiempo Quicksort
    quicksort_time = timeit.timeit(lambda: quicksort(arr.copy()), number=1)
    
    print(f"Tamaño: {size} | Burbuja: {bubble_time:.6f}s | Quicksort: {quicksort_time:.6f}s")
```

# 5. Visualización gráfica 

```python
import matplotlib.pyplot as plt

sizes = [100, 500, 1000, 5000]
bubble_times = []
quicksort_times = []

for size in sizes:
    arr = [random.randint(0, 10000) for _ in range(size)]
    bubble_times.append(timeit.timeit(lambda: bubble_sort(arr.copy()), number=1))
    quicksort_times.append(timeit.timeit(lambda: quicksort(arr.copy()), number=1))

plt.plot(sizes, bubble_times, label="Burbuja", marker='o', linestyle="--")
plt.plot(sizes, quicksort_times, label="Quicksort", marker='o', linestyle="--")
plt.xlabel("Tamaño del arreglo")
plt.ylabel("Tiempo (s)")
plt.title("Comparación de Algoritmos de Ordenamiento")
plt.legend()
plt.show()
```

## **Análisis de Resultados**

1. **El algoritmo de Burbuja tiene una complejidad de \(O(n^2)\), lo que lo hace ineficiente para grandes volúmenes de datos.**
2. **Quicksort tiene una complejidad promedio de \(O(n \log n)\), por lo que es significativamente más rápido.**
3. **Para pequeños volúmenes de datos (100-500 elementos), la diferencia no es tan evidente, pero en 5000 elementos, Quicksort es hasta 100 veces más rápido.**
